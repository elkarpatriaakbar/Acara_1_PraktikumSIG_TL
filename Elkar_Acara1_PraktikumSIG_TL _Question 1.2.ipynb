## Functions

Struktur kontrol yang telah kita pelajari sebelumnya (seperti *for loop* dan *if–else*) dapat disusun dan digunakan kembali dalam bentuk **function** (fungsi). Lalu, apa yang membedakan function dengan *for loop* biasa?

**Function** adalah sekumpulan kode yang disimpan di dalam memori (mirip dengan *object* yang telah dibahas pada Minggu 1), tetapi **tidak akan dijalankan sebelum dipanggil**. Artinya, Anda dapat mendefinisikan sebuah function di awal atau tengah program, lalu menggunakannya (memanggilnya) di bagian lain dari kode sesuai kebutuhan.

Pada umumnya, function menerima **parameter**, yaitu nilai masukan (input) atau argumen yang digunakan untuk menyesuaikan cara kerja kode di dalam function tersebut. Dengan parameter, satu function dapat digunakan untuk berbagai kondisi yang berbeda.

Dalam Python:
- Kata kunci `def` digunakan untuk **mendefinisikan function**
- Perintah `return` biasanya digunakan di akhir function untuk **mengembalikan hasil** dari function (jika ada)

Contoh sintaks dasar function:

```python
def fungsiBaru(param1, param2):


````python
ls = [1,2,3,4,5,6,7,8,9]

def ambil_elemen_pertama(daftar):  # mendefinisikan function
    return daftar[0]               # mengembalikan elemen pertama dari list

ambil_elemen_pertama(ls)           # memanggil function
````

Berhentilah sejenak dan perhatikan logika dari function yang akan kita buat di bawah ini, yaitu `nilaiMean_list`.  
* Apa makna penggunaan pernyataan `for` di dalam function tersebut?  
* Apa arti operator `+=`? Jika belum yakin, cobalah cari di Google untuk memahami maksudnya.

````python
ls = [1, 2, 3, 4, 5, 6, 7, 8]

# Di sini kita akan mendefinisikan sebuah fungsi bernama nilaiMean_list
def nilaiMean_list(ls1):
    jumlah = 0                  # variabel untuk menyimpan total nilai
    for i in ls1:               # perulangan untuk setiap elemen dalam list
        jumlah += i             # menambahkan nilai elemen ke total
    panjang = len(ls1)          # menghitung jumlah elemen dalam list
    rata_rata = jumlah / panjang  # menghitung nilai rata-rata
    return rata_rata             # mengembalikan hasil rata-rata

print(nilaiMean_list(ls))       # memanggil fungsi dan menampilkan hasilnya
````

#### 😉 KUIS

Bisakah Anda membuat sebuah fungsi baru yang dapat mengembalikan jumlah data(count), total nilai (sum), nilai rata-rata (mean), dan nilai tengah (median)?
Gantilah tanda ??? dengan jawaban Anda.

````python
def get_statistics(data):
    # Menghitung jumlah elemen dalam list
    count = len(data)
    
    # Menghitung total seluruh nilai dalam list
    total = sum(data)
    
    # Menghitung nilai rata-rata (mean)
    # Rumus mean = total nilai / jumlah data
    mean = total / count  
    
    # Mengurutkan data terlebih dahulu untuk perhitungan median
    sorted_data = sorted(data)
    
    # Mengecek apakah jumlah data ganjil atau genap
    if count % 2 == 1:
        # Jika ganjil:
        # Median adalah elemen yang berada tepat di tengah
        median = sorted_data[count // 2]
    else:
        # Jika genap:
        # Median adalah rata-rata dari dua elemen tengah
        mid1 = sorted_data[count // 2 - 1]  # elemen tengah kiri
        mid2 = sorted_data[count // 2]      # elemen tengah kanan
        median = (mid1 + mid2) / 2
    
    # Mengembalikan 4 value: count, total, mean, median
    return count, total, mean, median


# Contoh penggunaan fungsi
data = [1, 5, 4, 3]

# Memanggil fungsi dan menyimpan hasilnya
hasil = get_statistics(data)

# Menampilkan hasil
print("Count :", hasil[0])   # jumlah data
print("Sum   :", hasil[1])   # total nilai
print("Mean  :", hasil[2])   # rata-rata
print("Median:", hasil[3])   # nilai tengah
````

## Level Advanced: Class dan Object dalam Python

Python merupakan bahasa pemrograman yang berbasis **object-oriented programming (OOP)**. Dalam Python, **objek** didefinisikan menggunakan **class**. Pada dasarnya, hampir semua hal di Python adalah objek. Class digunakan untuk mendefinisikan struktur data baru yang dapat disesuaikan dengan kebutuhan aplikasi yang dibuat.

Untuk membuat sebuah class baru, digunakan kata kunci `class`.

```python
class MyClass:
```

Selain itu, fungsi-fungsi yang berada di dalam atau “melekat” pada suatu objek disebut **method**. *Method* dapat diakses menggunakan tanda titik (.) setelah objek tersebut.

Sebagai contoh, kode berikut digunakan untuk mengubah seluruh huruf dalam sebuah string menjadi huruf kapital:

```console
"hello world!".upper()
```

````python
class Kucing:
    def __init__(self, nama):
        # atribut nama diisi saat objek kucing dibuat
        self.nama = nama  

    # atribut yang sama untuk semua objek Kucing
    jenis = "kucing domestik"
    warna = "putih"

    # method (perilaku) kucing
    def makan(self):
        print(self.nama + " sedang makan.")

    def suara(self):
        print(self.nama + " mengeong.")
````

````python
kucing1 = Kucing("Mimi")
kucing1.makan()
kucing1.suara()
````

## Namespace

Dalam Python (seperti halnya beberapa bahasa pemrograman lain), **namespace** adalah sekumpulan tanda atau "nama" yang digunakan untuk mengidentifikasi dan merujuk ke berbagai macam objek. 

Sebuah *namespace* memastikan bahwa semua objek yang telah didefinisikan memiliki nama yang unik, sehingga memudahkan proses identifikasi. Bayangkan *namespace* seperti sebuah **Buku Telepon**: meskipun ada banyak orang bernama "Budi" di kota lain, di dalam satu rumah (namespace) tertentu, nama "Budi" hanya merujuk pada satu orang yang spesifik. Untuk membaca lebih lanjut bisa cek [di sini](https://medium.com/@gough.cory/understanding-python-namespaces-through-recursion-c921cbd4e522).

### Jenis-Jenis Namespace:
* **Built-in:** Nama-nama bawaan Python yang selalu tersedia (contoh: `print`, `len`).
* **Global:** Nama yang didefinisikan di tingkat utama program/script.
* **Local:** Nama yang hanya ada di dalam sebuah fungsi (bersifat sementara).

Konsep ini sangat mirip dengan **Environments** (global/local) yang ada pada bahasa pemrograman **R**.

### Visualisasi Struktur Namespace (Aturan LEGB)

Python mencari nama variabel menggunakan urutan dari lingkup terkecil ke terbesar:

1.  **Lokal (Local):** Di dalam fungsi atau *lambda*.
2.  **Enclosing:** Di dalam fungsi yang membungkus fungsi lain (*nested functions*).
3.  **Global:** Di tingkat atas modul/file.
4.  **Built-in:** Fungsi bawaan Python.

## Python Package Manager – Pip, Conda, dan Mamba

Ada tiga package manager yang umum digunakan dalam ekosistem Python, yaitu pip, conda, dan mamba. Ketiganya digunakan untuk menginstal dan mengelola library, tetapi memiliki karakteristik dan kegunaan yang sedikit berbeda.

#### Pip
Pip adalah alat resmi yang direkomendasikan oleh Python Packaging Authority untuk menginstal paket dari Python Package Index (PyPI).
Pip menginstal paket dalam bentuk wheel atau source distribution. Untuk beberapa paket, terutama yang berbasis source, sistem mungkin memerlukan compiler dan library pendukung agar instalasi berhasil.

Contoh penggunaan:
```console
!pip install nama_paket
```

#### Conda
Conda adalah package manager sekaligus environment manager lintas platform yang dikembangkan oleh Anaconda. Conda dapat menginstal paket dari Anaconda Repository maupun Anaconda Cloud.

Keunggulan utama conda:

* Paket berbentuk binary (tidak memerlukan compiler),
* Dapat mengelola virtual environment,
* Tidak hanya untuk Python (juga mendukung C/C++, R, dan software lain).

Conda sangat berguna ketika:

* Membutuhkan lingkungan Python yang terpisah,
* Menghadapi konflik versi antar library.

#### Mamba
Mamba adalah alternatif untuk conda dengan fungsi hampir sama, tetapi jauh lebih cepat.
Mamba ditulis menggunakan bahasa C++ dan menggunakan dependency solver yang lebih efisien.

Secara sederhana:
* Conda → stabil dan umum digunakan
* Mamba → conda versi lebih cepat

Perintah mamba sama persis dengan conda, hanya kata conda diganti menjadi mamba.

````python
# a) Buat sebuah fungsi bernama `Olimpiade()`. Fungsi ini menerima input berupa tahun (antara 2000 sampai 2021), lalu menampilkan kota tempat Olimpiade  diselenggarakan pada tahun tersebut. 
# Jika pada tahun itu tidak ada Olimpiade, tampilkan pesan bahwa Olimpiade tidak diselenggarakan.
def Olimpiade(tahun):
    # Dictionary berisi data Olimpiade (Summer Olympics) 2000–2021
    data_olimpiade = {
        2000: "Sydney",
        2004: "Athena",
        2008: "Beijing",
        2012: "London",
        2016: "Rio de Janeiro",
        2021: "Tokyo"   # seharusnya 2020, tapi diselenggarakan 2021 (COVID-19)
    }
    
    # Mengecek apakah tahun ada dalam dictionary
    if tahun in data_olimpiade:
        print(f"Olimpiade diselenggarakan di {data_olimpiade[tahun]} pada tahun {tahun}.")
    else:
        print(f"Olimpiade tidak diselenggarakan pada tahun {tahun}.")
        

# Contoh penggunaan
# Menampilkan semua tahun dari 2000 sampai 2021
for tahun in range(2000, 2022):   # 2022 karena range stop itu eksklusif
    Olimpiade(tahun)
````

````python
# b) Buat sebuah fungsi bernama `ls_info()`. Fungsi ini menerima sebuah list yang berisi angka acak, kemudian menampilkan: jumlah elemen dalam list, nilai terbesar, dan nilai terkecil dari list tersebut.
def ls_info(data):
    # Menghitung jumlah elemen
    jumlah = len(data)
    
    # Mencari nilai terbesar
    terbesar = max(data)
    
    # Mencari nilai terkecil
    terkecil = min(data)
    
    # Menampilkan hasil
    print("Jumlah elemen :", jumlah)
    print("Nilai terbesar:", terbesar)
    print("Nilai terkecil:", terkecil)


# Contoh penggunaan
contoh_list = [47, 10, 98, 32, 65, 30]
ls_info(contoh_list)
````

````python
#  c) Buat sebuah fungsi untuk mencari nilai median dari sebuah list angka yang belum diurutkan. (Petunjuk: urutkan list terlebih dahulu sebelum menghitung median)
def cari_median(data):
    # Mengurutkan list terlebih dahulu
    sorted_data = sorted(data)
    
    n = len(sorted_data)
    
    # Jika jumlah data ganjil
    if n % 2 == 1:
        median = sorted_data[n // 2]
    else:
        # Jika jumlah data genap
        mid1 = sorted_data[n // 2 - 1]
        mid2 = sorted_data[n // 2]
        median = (mid1 + mid2) / 2
    
    return median


# Contoh penggunaan
data_median = [7, 3, 9, 5, 1]
print("Median:", cari_median(data_median))
````
