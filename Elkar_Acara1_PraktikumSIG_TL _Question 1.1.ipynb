## Operator Dasar Python

Operator-operator ini digunakan untuk mengolah dan mengubah nilai antar variabel, atau untuk melakukan perhitungan yang melibatkan lebih dari satu objek atau elemen dalam daftar data numerik.

**Operator Pemberian Nilai (Assignment):** `=`

**Operator Aritmatika:** ```+ , - , * , / , ** , %```

**Operator Perbandingan (Comparison):** ```== , != , < , <= , > , >=```

Perhatikan bahwa untuk membandingkan dua nilai (misalnya menyatakan kondisi ketika x **sama dengan** y), kita menggunakan **dua tanda sama dengan (==)**, bukan satu tanda sama dengan (`=`) yang digunakan untuk memberi nilai pada variabel.

**Operator Logika (Logical):** ```and , or```

````python
# operator aritmatika
x = 5
y = 6
z = x + y

print(z)
````

````python
# operator perbandingan
print(z == x)  # menghasilkan nilai TRUE atau FALSE
````

````python
# operator logika
x < y or y < z  # menghasilkan TRUE jika kondisi terpenuhi; FALSE jika tidak
````

Selain digunakan untuk menjumlahkan nilai numerik, tanda `+` juga dapat digunakan untuk **menggabungkan (concatenate)** satu atau lebih string, atau string yang disimpan sebagai objek dalam Python.

````python
# operator aritmatika tidak dapat digunakan pada string
x = "Model 1 accuracy:"
y = "100%"

# kecuali operator `+`
print(x + ' ' + y)
````

### Dictionary

Dictionary adalah kumpulan data yang tidak berurutan yang disimpan dalam bentuk **pasangan kunci:nilai** (key:value). Berbeda dengan sequence yang diindeks oleh nomor urut, dictionary diindeks menggunakan kunci (key), yang dapat berupa tipe data immutable: string dan angka selalu bisa digunakan sebagai key.

Contohnya, kita bisa menggunakan `ls[0]` untuk mengakses elemen pertama dari list `ls`, tetapi pada dictionary kita memerlukan key yang sesuai untuk mengakses suatu nilai. Dictionary didefinisikan menggunakan tanda kurung kurawal `{}`.

```console
contoh_dict = {"kunci1": 10}
```

Hanya objek **immutable** yang bisa digunakan sebagai key, sehingga list tidak bisa digunakan.
Contoh valid (tuple sebagai key):
```console
contoh_dict = {('hello', 'world'): True}  # valid - tuple bisa digunakan sebagai key dictionary
```

Contoh tidak valid (list sebagai key):
```console
contoh_dict = {['hello', 'world']: True}  # invalid - list tidak bisa digunakan sebagai key
```


````python
# dictionary didefinisikan menggunakan {key:value} atau dict()
d1 = {}  # membuat objek dictionary kosong
d1['nama'] = "Budi"
d1['umur'] = 25
d1['lokasi'] = "Sleman"
print(d1)
````

````python
# mengakses elemen dalam dictionary
print(d1['nama'])
````

````python
# membuat dictionary dengan melakukan enumerasi pada list
# Enumerasi adalah proses memberikan **nomor urut (index)** pada setiap elemen dalam list, sehingga kita bisa mengaksesnya secara berpasangan (index, nilai)
newDict = {}  # membuat objek dictionary kosong
list1 = ['budi', 'sasa', 'steven', 'kevin', 'micky', 'pras', 'andrew']  # membuat list string yang akan dimasukkan ke dictionary

for i, j in enumerate(list1):  # menggunakan enumerate, kita mendapatkan pasangan (index, nilai) dari list1 dan memasukkannya ke dictionary
    newDict[i] = j

newDict  # menampilkan objek dictionary yang baru dibuat
````

### Set

Set adalah kumpulan elemen yang tidak berurutan dan hanya berisi elemen unik. Set juga didefinisikan menggunakan tanda kurung kurawal `{}`.
```console
contoh_set = {"sasa", "ade", "sasa"}
```

Kita bisa menambahkan `(add)` atau menghapus `(remove)` elemen dari sebuah set setelah set tersebut dibuat. Namun, **elemen** di dalam set harus bersifat **immutable**, misalnya angka, string, atau tuple. Anda tidak bisa memasukkan list atau dictionary ke dalam set.

````python
my_set = {1, 2, 3}
my_set.add(4)       # menambahkan elemen 4
my_set.remove(2)    # menghapus elemen 2
print(my_set)       # output: {1, 3, 4}
````

````python
# set didefinisikan menggunakan {} atau set()
set1 = {1, 1, 1, 1, 10}  # elemen duplikat akan otomatis dihapus
set1
````

````python
# operasi pada set

set1 = {1, 1, 1, 1, 10}
set2 = {1, 1, 1, 1, 20}

print(set1.intersection(set2))  # irisan: elemen yang ada di kedua set
print(set1.union(set2))         # gabungan: semua elemen dari kedua set
print(set1.difference(set2))    # selisih: elemen yang ada di set1 tapi tidak di set2
````

### Range

```console
range(1, 10, 1)
```

````python
# range
k = range(1, 10, 2)
k[1]  # mengakses elemen kedua dari range
````

## Control Flow

Struktur kontrol mirip dengan blok program yang menganalisis variabel dan memilih jalur eksekusi berdasarkan parameter tertentu yang biasanya ditentukan oleh pengguna. Struktur ini merepresentasikan proses pengambilan keputusan dasar dalam pemrograman dan dapat diterapkan melalui tiga tipe dasar, yaitu:

* Urutan (Sequence) → mengeksekusi kode satu per satu.

* Percabangan (Branching/Selection) → memilih jalur eksekusi berdasarkan kondisi tertentu.

* Perulangan (Repetition/Looping) → menjalankan kode berulang kali.

Contoh struktur kontrol yang akan dipelajari meliputi:

### If
Pernyataan `if` digunakan untuk mengambil keputusan dalam program.
* Jika kondisi `True`, kode di dalam if dijalankan.
* Jika kondisi `False`, kode di dalam if tidak dijalankan.

Kita bisa menambahkan `else` untuk menentukan apa yang terjadi jika kondisi salah.

````python
nilai = 80

if nilai >= 75:
    print("Selamat, kamu lulus!")
else:
    print("Maaf, kamu belum lulus.")
````

````python
x = 8
y = 9
z = 10

# beberapa kondisi (elif)
if x > y:
    print(True)
elif y > z:
    print(True)
else:
    print(False)
````

### For

Pernyataan `for` digunakan untuk mengulang kode beberapa kali untuk setiap elemen dalam sebuah daftar, string, atau range.
Contohnya:
* Ingin mencetak semua nama dalam daftar.
* Ingin menjumlahkan angka dari 1 sampai 10.

````python
# list sebagai objek yang bisa diiterasi
ls1 = [11, 12, 13, 14, 15, 16]

# mencetak semua elemen dalam list
for i in ls1:
    print(i)
````

````python
# string sebagai objek yang bisa diiterasi
text = "hello world"

# mencetak setiap karakter dalam string secara terpisah
for i in text:
    print(i)
````

````python
# membuat iterator menggunakan range
for i in range(0, 10, 1):
    print(i)
````

### While

Pernyataan `while` digunakan untuk mengulang kode selama suatu kondisi bernilai `True`.
* Kode akan terus dijalankan selama kondisi terpenuhi.
* Jika kondisi menjadi `False`, perulangan berhenti.

````python
### While loop - menghitung mundur
jumlah = 5

while jumlah > 0:
    print("Hitung mundur:", jumlah)
    jumlah -= 1  # mengurangi nilai jumlah 1 setiap iterasi

print("Waktu habis!")
````

**f)** Buat fungsi sederhana `get_median()` untuk menghitung **median** dari sebuah list bilangan bulat. Uji fungsi ini pada list `ls` dan `lx`. *Petunjuk: median dihitung berbeda jika jumlah elemen list genap atau ganjil*. <br />

````python
# a) Membuat list ls dari 1 sampai 20 (inklusif)
ls = list(range(1, 21))
````

````python
# b) Cetak elemen pertama, keempat, dan kesepuluh
print("Elemen pertama:", ls[0])
print("Elemen keempat:", ls[3])
print("Elemen kesepuluh:", ls[9])
````

````python
# c) Cetak elemen ketiga dari belakang
print(f"Elemen ketiga dari belakang pada list 'ls' adalah {ls[-3]}.")
````

````python
# d) Cetak bilangan genap dalam ls
bilangan_genap = [x for x in ls if x % 2 == 0]
print(f"Bilangan genap dalam list 'ls' adalah {bilangan_genap}.")
````

````python
# e) Membuat list baru lx dari ls hingga 49
lx = ls.copy()
for i in range(21, 50):
    lx.append(i)
````

````python
# f) Fungsi untuk menghitung median
def get_median(data):
    n = len(data)
    sorted_data = sorted(data)
    
    if n % 2 == 1:  # jika jumlah elemen ganjil
        return sorted_data[n // 2]
    else:  # jika jumlah elemen genap
        mid1 = sorted_data[n // 2 - 1]
        mid2 = sorted_data[n // 2]
        return (mid1 + mid2) / 2

# Uji fungsi median
print("Median ls:", get_median(ls))
print("Median lx:", get_median(lx))
````

````python
# g) Membuat lx_copy dari ls, kelipatan 5 diganti 'x'
lx_copy = []
for x in ls:
    if x % 5 == 0:
        lx_copy.append('x')
    else:
        lx_copy.append(x)

print("lx_copy:", lx_copy)
````

* Jika elemennya genap, hitung pangkat dua (kuadratnya) <br />
Simpan seluruh hasil perhitungan tersebut ke dalam sebuah list baru. (Petunjuk: Gunakan list kosong dan metode append()). <br />

**d)** Buat sebuah list baru bernama `ls3`. Setiap elemen `ls3` merupakan hasil penjumlahan: *elemen ke-i dari awal* `ls1`, dan *elemen ke-i dari akhir* `ls2`. <br />
Sebagai contoh: <br />

**e)** Buat sebuah list bernama `lx` yang berisi 20 bilangan bulat acak dengan nilai antara 50 sampai 100. (Petunjuk: Gunakan modul *random*). <br />

**f)** Urutkan elemen-elemen pada list `lx` dari nilai terbesar ke terkecil, lalu simpan hasilnya ke dalam list baru bernama `lx2`.<br />

**g)** Buat sebuah program yang menerima sebuah list, kemudian menghasilkan list baru dengan urutan elemen yang dibalik. Sebagai contoh, jika input adalah [1, 5, 4, 3], maka output yang dihasilkan adalah [3, 4, 5, 1]. <br />

````python
# a) Membuat ls1 dan ls2
ls1 = list(range(1, 21))          # 1 sampai 20
ls2 = list(range(12, 51, 2))      # genap dari 12 sampai 50

print("ls1:", ls1)
print("ls2:", ls2)
````

````python
# b) Menjumlahkan elemen pada indeks genap (0,2,4,...) menggunakan for dan if
jumlah_ls1 = 0
jumlah_ls2 = 0

for i in range(len(ls1)):
    if i % 2 == 0:
        jumlah_ls1 += ls1[i]

for i in range(len(ls2)):
    if i % 2 == 0:
        jumlah_ls2 += ls2[i]

print("Jumlah elemen indeks genap ls1:", jumlah_ls1)
print("Jumlah elemen indeks genap ls2:", jumlah_ls2)
````

````python
# c) Proses ls1:
# - jika ganjil → akar kuadrat
# - jika genap → kuadrat
import math

hasil_ls1 = []

for x in ls1:
    if x % 2 == 1:   # ganjil
        hasil_ls1.append(math.sqrt(x))
    else:            # genap
        hasil_ls1.append(x ** 2)

print("Hasil proses ls1:", hasil_ls1)
````

````python
# d) Membuat ls3 = elemen ke-i ls1 + elemen ke-i dari akhir ls2
ls3 = []

for i in range(len(ls1)):
    ls3.append(ls1[i] + ls2[-(i+1)])

print("ls3:", ls3)
````

````python
# e) Membuat lx berisi 20 bilangan acak antara 50-100
import random

lx = []
for _ in range(20):
    lx.append(random.randint(50, 100))

print("lx:", lx)
````

````python
# f) Mengurutkan lx dari terbesar ke terkecil
lx2 = sorted(lx, reverse=True)

print("lx2 (descending):", lx2)
````

````python
# g) Program membalik urutan list
def reverse_list(input_list):
    hasil = []
    for i in range(len(input_list)-1, -1, -1):
        hasil.append(input_list[i])
    return hasil

# Contoh uji
contoh = [1, 5, 4, 3]
print("Reverse contoh:", reverse_list(contoh))
````
